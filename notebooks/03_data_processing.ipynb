```python
import pandas as pd
```

```python
# KR (full)
kr_df = pd.read_stata(
    "C:\\Users\\Arsh\\College\\Semester 5\\INT - 396\\RHPA-India\\data\\raw\\kr_data.dta",
    convert_categoricals=False
)

# IR (chunked)
reader = pd.read_stata(
    "C:\\Users\\Arsh\\College\\Semester 5\\INT - 396\\RHPA-India\\data\\raw\\ir_data.dta",
    convert_categoricals=False,
    chunksize=100000
)

ir_df = next(reader)   # only first chunk for now
```

```python
ir_cols = [
    'sdist',
    'v106', 'v190',
    'v113', 'v116', 'v119',
    'v171a', 'v157', 'v158', 'v159',
    'v463a', 'v463b'
]

kr_cols = [
    'sdist',
    'hw70', 'hw71', 'hw72'
]

ir_df = ir_df[ir_cols]
kr_df = kr_df[kr_cols]
```

```python
ir_df = ir_df.dropna()
kr_df = kr_df.dropna()
```

```python
# Replace invalid values with NaN
kr_df[['hw70', 'hw71', 'hw72']] = kr_df[['hw70', 'hw71', 'hw72']].replace(
    [9996, 9997, 9998, 9999],
    pd.NA
)
```

```python
kr_df = kr_df.dropna(subset=['hw70', 'hw71', 'hw72'])
```

```python
kr_df['health_score'] = (
    kr_df['hw70'] +
    kr_df['hw71'] +
    kr_df['hw72']
) / 3
```

```python
kr_df['health_score'] = pd.to_numeric(kr_df['health_score'], errors='coerce')
```

```python
kr_df = kr_df.dropna(subset=['health_score'])
```

```python
ir_df['sanitation_score'] = (
    ir_df['v113'] +
    ir_df['v116'] +
    ir_df['v119']
) / 3
```

```python
ir_df['education_score'] = (
    ir_df['v106'] +
    ir_df['v190']
) / 2
```

```python
ir_df['awareness_score'] = (
    ir_df['v171a'] +
    ir_df['v157'] +
    ir_df['v158'] +
    ir_df['v159']
) / 4
```

```python
ir_df['behavior_score'] = (
    ir_df['v463a'] +
    ir_df['v463b']
) / 2
```
